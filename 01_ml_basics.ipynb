- Machine Learning: A model learns patterns from sample data.
- Deep Learning: A machine learning approach that utilizes multi-layered artificial neural networks.
- Supervised Learning: Data includes target labels—for example, labeling a photograph as either a cat or a dog.
- Unsupervised Learning: Searching for structures or patterns within data that lacks target labels.
- Classification: Assigning data to one of several predefined categories.
- Data Leakage: The inclusion of information that should not be accessible during training, which leads to misleading evaluation results.

```python

import torch

# A simple dataset consisting of 20 samples
data = torch.arange(20)

# To ensure the same random split can be reproduced
generator = torch.Generator().manual_seed(42)
indices = torch.randperm(len(data), generator=generator)

# 70% train, 15% validation, 15% test
train_data = data[indices[:14]]
val_data = data[indices[14:17]]
test_data = data[indices[17:]]

print("Train:", train_data.tolist())
print("Validation:", val_data.tolist())
print("Test:", test_data.tolist())

print("Total:", len(train_data) +
len(val_data) + len(test_data))
```
